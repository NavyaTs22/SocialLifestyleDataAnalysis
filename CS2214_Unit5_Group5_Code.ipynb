# EXPLORATORY DATA ANALYSIS


```python
# Create a copy of the original dataset
df_clean = df.copy()

# Remove duplicate rows
df_clean = df_clean.drop_duplicates()

print("Rows after removing duplicates:", df_clean.shape[0])
```

```python
# Fill missing numerical values with median
for column in df_clean.select_dtypes(include='number').columns:
    df_clean[column] = df_clean[column].fillna(df_clean[column].median())

# Fill missing categorical values with mode
for column in df_clean.select_dtypes(include='object').columns:
    if df_clean[column].isnull().sum() > 0:
        df_clean[column] = df_clean[column].fillna(df_clean[column].mode()[0])

print("Missing values handled successfully.")
```

```python
# Checked whether any missing values remain after cleaning
print("Missing Values:")
print(df_clean.isnull().sum())
```

```python
# Checked whether any duplicate rows remain after cleaning
print("Remaining duplicate rows:", df_clean.duplicated().sum())
```

```python
# Detect and remove outliers using IQR method

outlier_columns = [
    'daily_social_media_time',
    'sleep_hours',
    'screen_time_before_sleep'
]

df_no_outliers = df_clean.copy()

for column in outlier_columns:

    Q1 = df_no_outliers[column].quantile(0.25)
    Q3 = df_no_outliers[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    df_no_outliers = df_no_outliers[
        (df_no_outliers[column] >= lower_limit) &
        (df_no_outliers[column] <= upper_limit)
    ]

print("Original dataset shape:", df_clean.shape)
print("Dataset after outlier removal:", df_no_outliers.shape)
print("Number of rows removed:", len(df_clean) - len(df_no_outliers))
```

```python
# Import MinMaxScaler

from sklearn.preprocessing import MinMaxScaler

# Create a copy of the dataset

df_scaled = df_no_outliers.copy()

# Select numerical columns

numeric_columns = df_scaled.select_dtypes(include='number').columns

# Apply MinMax scaling

scaler = MinMaxScaler()

df_scaled[numeric_columns] = scaler.fit_transform(
    df_scaled[numeric_columns]
)

# Display the scaled dataset

print("Scaled dataset:")
print(df_scaled.head())
```

# DATA VISUALIZATION

```python
# Created a histogram to show the distribution of daily social media usage
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df_clean,
    x='daily_social_media_time',
    bins=20,
    kde=True
)

plt.title('Distribution of Daily Social Media Usage')
plt.xlabel('Daily Social Media Time (Hours)')
plt.ylabel('Number of Records')

plt.tight_layout()
plt.savefig('chart1_social_media_distribution.png')
plt.show()
```

###Interpretation:                                                   The histogram shows how daily social media screen time is distributed across the sample. It highlights whether usage centers around moderate duration or exhibits a skewed distribution toward higher screen time hours, providing baseline insight into digital habits.

```python
# Calculate and visualize average productivity across different job types
productivity_by_job = df_clean.groupby(
    'job_type'
)['actual_productivity_score'].mean().sort_values()

plt.figure(figsize=(9, 5))

sns.barplot(
    x=productivity_by_job.index,
    y=productivity_by_job.values
)

plt.title('Average Productivity by Job Type')
plt.xlabel('Job Type')
plt.ylabel('Average Actual Productivity Score')
plt.xticks(rotation=30)

plt.tight_layout()
plt.savefig('chart2_productivity_by_job.png')
plt.show()
```

###Interpretation:                                                The bar chart ranks job types by their average actual productivity scores. Visualizing the sorted values helps identify which occupational roles report higher output and whether specific job types experience reduced productivity levels.

```python
# Create a scatter plot to examine the relationship between social media usage and productivity
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df_clean,
    x='daily_social_media_time',
    y='actual_productivity_score',
    alpha=0.4
)

plt.title('Social Media Usage vs Actual Productivity')
plt.xlabel('Daily Social Media Time (Hours)')
plt.ylabel('Actual Productivity Score')

plt.tight_layout()
plt.savefig('chart3_social_media_vs_productivity.png')
plt.show()
```

### Interpretation:                                                The scatter plot examines the correlation between daily social media hours and actual productivity scores. The semi-transparent markers (alpha=0.4) allow observation of data density, showing whether increased screen time relates to lower productivity or if output remains variable across usage levels.

```python
# Create a box plot to compare stress levels across different job types
plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df_clean,
    x='job_type',
    y='stress_level'
)

plt.title('Stress Level Distribution Across Job Types')
plt.xlabel('Job Type')
plt.ylabel('Stress Level')
plt.xticks(rotation=30)

plt.tight_layout()
plt.savefig('chart4_stress_by_job.png')
plt.show()
```

### Interpretation:                                                The box plot displays the median, interquartile range (IQR), and potential outliers in stress levels across different job roles. This visual highlights variations in stress spread and baseline pressure across occupations.
